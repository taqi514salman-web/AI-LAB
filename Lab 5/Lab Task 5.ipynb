```python
import math


# ------------------------------------------------------------
# Minimax without pruning
# ------------------------------------------------------------

def minimax(depth, index, max_turn, scores, target_depth, counter):
    # A leaf node contains the final score of the game.
    if depth == target_depth:
        counter[0] = counter[0] + 1
        return scores[index]

    if max_turn:
        left_value = minimax(depth + 1, index * 2, False,
                             scores, target_depth, counter)
        right_value = minimax(depth + 1, index * 2 + 1, False,
                              scores, target_depth, counter)
        return max(left_value, right_value)
    else:
        left_value = minimax(depth + 1, index * 2, True,
                             scores, target_depth, counter)
        right_value = minimax(depth + 1, index * 2 + 1, True,
                              scores, target_depth, counter)
        return min(left_value, right_value)


# ------------------------------------------------------------
# Minimax with alpha-beta pruning
# ------------------------------------------------------------

def alphabeta(depth, index, max_turn, scores, target_depth,
              alpha, beta, counter):
    # A leaf node contains the final score of the game.
    if depth == target_depth:
        counter[0] = counter[0] + 1
        return scores[index]

    if max_turn:
        value = float('-inf')

        # Visit the left branch first, followed by the right branch.
        for child in range(2):
            child_index = index * 2 + child
            child_value = alphabeta(depth + 1, child_index, False,
                                    scores, target_depth, alpha, beta,
                                    counter)
            value = max(value, child_value)
            alpha = max(alpha, value)

            # No MIN ancestor will allow the remaining branch.
            if alpha >= beta:
                break

        return value
    else:
        value = float('inf')

        # Visit the left branch first, followed by the right branch.
        for child in range(2):
            child_index = index * 2 + child
            child_value = alphabeta(depth + 1, child_index, True,
                                    scores, target_depth, alpha, beta,
                                    counter)
            value = min(value, child_value)
            beta = min(beta, value)

            # No MAX ancestor will allow the remaining branch.
            if alpha >= beta:
                break

        return value


# ------------------------------------------------------------
# Helper function for running both algorithms
# ------------------------------------------------------------

def run_test(test_name, scores):
    # The supplied manual uses a complete binary tree.
    target_depth = int(math.log2(len(scores)))

    minimax_counter = [0]
    alpha_beta_counter = [0]

    minimax_value = minimax(0, 0, True, scores, target_depth,
                            minimax_counter)
    alpha_beta_value = alphabeta(0, 0, True, scores, target_depth,
                                 float('-inf'), float('inf'),
                                 alpha_beta_counter)

    print("\n" + test_name)
    print("Scores:", scores)
    print("Minimax value:", minimax_value)
    print("Alpha-beta value:", alpha_beta_value)
    print("Minimax leaves evaluated:", minimax_counter[0])
    print("Alpha-beta leaves evaluated:", alpha_beta_counter[0])
    print("Leaves avoided:", minimax_counter[0] - alpha_beta_counter[0])
    print("Same optimal value:", minimax_value == alpha_beta_value)


# ------------------------------------------------------------
# Main program
# ------------------------------------------------------------

# Original complete tree from the manual's alpha-beta example.
original_scores = [3, 5, 2, 9, 3, 5, 2, 9]
run_test("Test 1: Original branch order", original_scores)

# The leaves inside each MAX node are swapped. This changes the order
# in which branches are searched, but it does not change the tree's
# Minimax value.
reordered_scores = [5, 3, 9, 2, 5, 3, 9, 2]
run_test("Test 2: Reordered branches", reordered_scores)

# Edge case: all terminal scores are equal.
# Both algorithms should return the same value.
equal_scores = [4, 4, 4, 4, 4, 4, 4, 4]
run_test("Test 3: Edge case with equal scores", equal_scores)


# ------------------------------------------------------------
# Hand-trace values for Test 1
# ------------------------------------------------------------

print("\nHand trace for Test 1")
print("MAX level: max(3, 5) = 5")
print("MAX level: max(2, 9) = 9")
print("MAX level: max(3, 5) = 5")
print("MAX level: max(2, 9) = 9")
print("MIN level: min(5, 9) = 5")
print("MIN level: min(5, 9) = 5")
print("Root MAX value: max(5, 5) = 5")
```
