```python
# Artificial Intelligence Lab 1
# Student: Muhammad Taqi Salman
# Roll No: 058
# Section: BSAI 3A


def check_eligibility(score, document_complete, attendance, late_days):

    failed_conditions = []

    # Condition 1: score must be at least 70
    if score < 70:
        failed_conditions.append("score")

    # Condition 2: documents must be complete
    if document_complete == False:
        failed_conditions.append("document completeness")

    # Additional condition 1: attendance must be at least 75%
    if attendance < 75:
        failed_conditions.append("attendance")

    # Additional condition 2: late submission must not be more than 2 days
    if late_days > 2:
        failed_conditions.append("submission timing")

    # If no condition failed, the student passes
    if len(failed_conditions) == 0:
        return "PASS - All four conditions passed."

    else:
        reason = "FAIL - Failed condition(s): "

        for condition in failed_conditions:
            reason = reason + condition + ", "

        # Remove the final comma and space
        reason = reason[:-2]

        return reason + "."


# Four test cases
# Format:
# Case name, score, document status, attendance, late days

test_cases = [
    ["Case 1 - all conditions pass", 85, True, 88, 0],
    ["Case 2 - score boundary failure", 69, True, 90, 0],
    ["Case 3 - added attendance failure", 80, True, 74, 0],
    ["Case 4 - added timing failure", 80, True, 80, 3]
]


print("LAB 1 RULE TEST OUTPUT")
print("Student: Muhammad Taqi Salman")
print("Roll No: 058")
print("Section: BSAI 3A")
print("-" * 60)


for test in test_cases:

    case_name = test[0]
    score = test[1]
    document_complete = test[2]
    attendance = test[3]
    late_days = test[4]

    result = check_eligibility(
        score,
        document_complete,
        attendance,
        late_days
    )

    print(case_name + ": " + result)
```

```python
# Lab 1: Compare rule-based and probability-based decisions  # Lab topic


# This function uses rules written directly by the programmer.
def rule_based(score, documents_complete):
    # Return Review when documents are complete AND score is at least 70.
    return "Review" if documents_complete and score >= 70 else "Hold"
    # If any condition is false, return Hold.


# This function uses a probability and a threshold value.
def probability_based(probability):
    # Return Review when probability is at least 0.70.
    return "Review" if probability >= 0.70 else "Hold"
    # If probability is less than 0.70, return Hold.


# Each tuple stores: case name, score, document status, and probability.
cases = [
    ("complete_high", 82, True, 0.81),   # Complete documents and high score
    ("complete_low", 68, True, 0.74),    # Complete documents but low score
    ("incomplete_high", 91, False, 0.88),# Incomplete documents but high score
    ("boundary", 70, True, 0.70),        # Values exactly equal to thresholds
]


# Print the headings for the output.
print("case | symbolic | threshold | disagreement")


# Repeat the following steps for every case in the list.
for name, score, complete, probability in cases:
    # Apply the rule-based decision method.
    symbolic_result = rule_based(score, complete)

    # Apply the probability-based decision method.
    threshold_result = probability_based(probability)

    # True means both methods gave different results.
    disagreement = symbolic_result != threshold_result

    # Print the case name and both results.
    print(name, "|", symbolic_result, "|", threshold_result, "|", disagreement)


# Print a heading before testing a new probability threshold.
print("\nChanging the probability threshold from 0.70 to 0.69:")


# Test every case again with the changed threshold.
for name, score, complete, probability in cases:
    # Calculate the result using the original threshold of 0.70.
    old_result = probability_based(probability)

    # Calculate the result using the new threshold of 0.69.
    new_result = "Review" if probability >= 0.69 else "Hold"

    # Print the probability and results from both thresholds.
    print(name, "| probability:", probability,
          "| old:", old_result, "| new:", new_result)
```
