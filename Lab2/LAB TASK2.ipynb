```python
# This class represents the classroom intelligent agent.
class ClassroomAgent:
    # This method receives temperature and occupancy as percepts.
    def act(self, temp, occupied):
        # If nobody is in the room, save energy.
        if not occupied:
            return "ECO"

        # If the occupied room is hot, turn on cooling.
        if temp > 26:
            return "COOL"

        # If the occupied room is cold, turn on warming.
        if temp < 20:
            return "WARM"

        # For temperatures from 20 to 26, keep the room idle.
        return "IDLE"


# This creates an object from the ClassroomAgent class.
agent = ClassroomAgent()

# Each tuple contains one percept: (temperature, occupied).
percepts = [(29, True), (24, True), (20, True), (26, True), (19, True), (29, False)]

# Print headings so the output is easy to read.
print("Step | Temperature | Occupied | Action")
print("-----|-------------|----------|-------")

# Send each percept to the agent and print its action.
for step, percept in enumerate(percepts, 1):
    # The star sends both tuple values as method arguments.
    action = agent.act(*percept)
    print(f"{step:>4} | {percept[0]:>11} | {str(percept[1]):>8} | {action}")
```

```python

# This class is the classroom agent with a small memory.
class ClassroomMemoryAgent:
    # The constructor runs when the object is created.
    def __init__(self):
        # At the beginning, there is no previous mode.
        self.previous_mode = None

    # This method receives temperature and occupancy as input.
    def decide(self, temp, occupied):
        # First choose the target mode from the current percept.
        if not occupied:
            target = "ECO"
        elif temp > 26:
            target = "COOL"
        elif temp < 20:
            target = "WARM"
        else:
            target = "IDLE"

        # Send a command only if the target mode has changed.
        if target == self.previous_mode:
            command = "NO_COMMAND"
        else:
            # Create a command for the new target mode.
            command = "SET_" + target

            # Store the current mode for the next decision.
            self.previous_mode = target

        # Return the selected mode and the command.
        return target, command


# This function runs one group of test percepts.
def run_test(name, percepts):
    # Create a new agent for each test.
    agent = ClassroomMemoryAgent()

    # Print the test name and table headings.
    print("\n" + name)
    print("Percept       Target  Command")
    print("------------- ------- ------------")

    # Give each percept to the agent and print the result.
    for percept in percepts:
        target, command = agent.decide(*percept)
        print(f"{str(percept):<13} {target:<7} {command}")


# Test 1 checks whether repeated hot readings repeat the command.
run_test(
    "Test 1: repeated hot reading",
    [(29, True), (29, True), (29, True)]
)

# Test 2 checks boundary temperatures and an empty room.
run_test(
    "Test 2: boundary and empty room",
    [(26, True), (20, True), (20, False)]
)

# Test 3 checks several changes in classroom conditions.
run_test(
    "Test 3: changing conditions",
    [(19, True), (21, True), (27, True), (27, False)]
)
```
